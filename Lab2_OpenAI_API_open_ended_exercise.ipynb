### <font color ='#0FAA82'>**💡Count Tokens Before Sending**

You can count tokens locally before making API calls to estimate costs.

````python
# Helper function to count tokens
def count_tokens(text: str, model: str = "gpt-4o-mini") -> int:
    """Count the number of tokens in a text string."""
    encoding = tiktoken.encoding_for_model(model)
    return len(encoding.encode(text))

# Example usage
test_text = "Write a blog post about cheese"
token_count = count_tokens(test_text)
print(f"The text '{test_text}' contains {token_count} tokens.")
````

### Let's try to alter some ***parameters*** like `max_tokens`, `teperature` and `top_p`

* `max_tokens`
  * Defines the maximum number of tokens the response can generate.

* `n` (Default: 1)
  * Defines the number of responses created.

* `temperature` (Default: 1)
  * This parameter helps you select the randomness in responses with values going from 0 to 2. If you select a value closer to 0, the responses will be less random, and a higher temperature will lead to more random responses.

* `top_p` (Default: 1)
  * This is another parameter that helps you decide the randomness using nucleus sampling.

>>*(Nucleus Sampling is a technique that cuts off lower probability tokens. For example, if you select top_p to be 0.3, the top 30% of tokens with highest probability will be considered for the responses.)*

<br>

> **Note:** It's not advised to change both temperature and top_p at the same time.


---

````python
prompt = "Write a blog post titled Cheese 101: What everyone and their brother should know about cheese" # @param {type:"string"}
````

````python
##Use the controls on the right to change the values

max_tokens = 100 # @param {type:"integer"}
temperature = 0.6 # @param {type:"slider", min:0, max:1, step:0.1}
top_p = 0.5 # @param {type:"slider", min:0, max:1, step:0.1}
````

````python
### Another example with Custom Parameters
## Try changing the teperature and run the cell again

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
      {"role": "system", "content": "You are a helpful writing assistant."},
      {"role": "user", "content": prompt}
        ],
    max_tokens=max_tokens,
    temperature=temperature,
    top_p=top_p
)

print(response.choices[0].message.content.strip())
````

## 💡 <font color ='#0FAA82'> **Streaming Responses**

For better user experience, especially with longer responses, you can stream the output as it's generated:


````python
print("Streaming response:\n")
print("=" * 50)

stream = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short poem about cheese."}
    ],
    max_tokens=100,
    stream=True  # Enable streaming
)

# Print each chunk as it arrives
for chunk in stream:
    if chunk.choices[0].delta.content is not None:
        print(chunk.choices[0].delta.content, end="", flush=True)

print("\n" + "=" * 50)
````

**Why streaming?**
- Provides immediate feedback to users
- Better UX for chatbots and interactive applications
- Allows users to start reading while generation continues

## 💡<font color ='#0FAA82'> **Structured Outputs & JSON Mode**

When you need structured, parseable data (critical for production apps), use JSON mode:


````python
# Using JSON mode to get structured output
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant that outputs in JSON."},
        {"role": "user", "content": "List 3 popular cheese types with their origin countries. Output as JSON with a 'cheeses' array."}
    ],
    response_format={"type": "json_object"},  # Ensure JSON output
    max_tokens=300
)

# Parse the JSON response
cheese_data = json.loads(response.choices[0].message.content)
print("Structured Cheese Data:")
print("=" * 50)
print(json.dumps(cheese_data, indent=2))
````

## 💡 <font color ='#0FAA82'> **Function Calling (Tools)**

Function calling lets LLMs interact with external systems and tools - fundamental for building AI agents:


````python
# Define a function/tool for the LLM to use
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_cheese_pairing",
            "description": "Get wine pairing recommendations for a specific cheese",
            "parameters": {
                "type": "object",
                "properties": {
                    "cheese_type": {
                        "type": "string",
                        "description": "The type of cheese (e.g., cheddar, brie, gouda)"
                      },
                    "intensity": {
                        "type": "string",
                        "enum": ["mild", "medium", "strong"],
                        "description": "The flavor intensity of the cheese"
                    }
                },
                "required": ["cheese_type"]
            }
        }
    }
]

# Make a request that should trigger function calling
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "What wine should I pair with aged cheddar?"}
    ],
    tools=tools,
    tool_choice="auto"  # Let the model decide when to use tools
)

# Check if the model wants to call a function
response_message = response.choices[0].message
print("Model's response:")
print("=" * 50)

if response_message.tool_calls:
    # The model wants to call a function
    tool_call = response_message.tool_calls[0]
    print(f"Function to call: {tool_call.function.name}")
    print(f"Arguments: {tool_call.function.arguments}")

    # In a real application, you would:
    # 1. Parse the arguments
    # 2. Call your actual function
    # 3. Send the result back to the model
    print("\nIn production, you would execute this function and return results to the model.")
else:
    print(response_message.content)
````

**Why function calling?**
- Enables LLMs to use external APIs and databases
- Foundation for building AI agents
- Provides structured way for models to request actions
- Makes LLMs actionable, not just conversational

## <font color ='#0FAA82'>**Use Embeddings to find similar text**




### **Instead of asking for a response we can even use the API to recieve word or sentence embeddings.**

### <font color ='#0F94AA'>**Maybe we can use embeddings to filter the first 100 comments from our cheese blog to find out who agrees with us 🤝** <font>

<center> <img src="https://drive.google.com/uc?export=download&id=1GKPeRmE0T2UagB3hqq4LS1izP7bQVkTF" width="250" height="200"> </center>


First, Let us compare this approach to Word2Vec and BERT:



- **Word2Vec (Static Embeddings)**: Generates embeddings based on the *local context* of words; however, it does not account for the broader sentence context or the varying meanings of words based on their usage.

- **BERT (Contextual Embeddings)**: Advances this by understanding words in *relation* to all other words in a sentence (bidirectional context), which allows it to capture nuanced meanings and dependencies.

- **OpenAI Embeddings (Sentence Level Contextual Embeddings)**: We will use the model "text-embedding-3-small" from Open AI. Similar to BERT, these are likely based on transformer architectures and provide sentence-level embeddings that consider the entire context of the input text, enhancing accuracy in tasks requiring a deep understanding of sentence meaning.

````python
# Creating the DataFrame with some sample reviews or comments
cheese_reviews = {
    "Name": [
        "Liam",
        "Jian",
        "Zoe",
        "Nikos",
        "Emma",
        "Sophia",
        "Xiaoyu",
        "Elena",
        "Dimitris",
        "Chloe",
        "Jia",
        "Michael",
        "Katie",
        "Daniel",
        "Oliver"
    ],
    "Review": [
        "The Gouda cheese from this place was the best I've ever tasted. The flavor was so rich!",
        "The blue cheese had an overwhelming, pungent aroma that made it inedible.",
        "The blue cheese I tried was disgusting. I think it was made from spoiled milk.",
        "The brie cheese was incredibly creamy and melted perfectly on the baguette.",
        "The aged Parmesan was perfectly sharp and added a great depth of flavor to the pasta.",
        "The ricotta cheese in the lasagna was light and fluffy, making the dish absolutely delicious.",
        "The Gruyère cheese in the French onion soup was beautifully melted and added great depth.",
        "The Monterey Jack cheese had a nice, mild flavor that complemented the quesadilla perfectly.",
        "I ordered a cheese platter but was given a grilled cheese sandwich instead. I ate it anyway.",
        "I hate Roquefort cheese. It is a disgrace. Somehow, it worked well in this salad though.",
        "Sharp cheddar cheese is a blessing to this world.",
        "The feta cheese in the Greek salad was too salty and ruined the dish.",
        "I tried the Manchego cheese and it had a delightful nutty flavor that complemented the wine.",
        "The cottage cheese was bland and had a strange texture. I wouldn't recommend it.",
        "I ate 11 slices of Swiss cheese and felt sick. The cheese was tasty in both directions."
    ]
}

#let's make pandas dataframe
df = pd.DataFrame(cheese_reviews)
````

````python
# Helper function to get embeddings

def get_embedding(txt):
    txt = txt.replace("\n", " ")
    response = client.embeddings.create(
        model="text-embedding-3-small",
        input=[txt]
    )
    return response.data[0].embedding

# Get embedding for each row of a text column of a DataFrame
df["embedding"] = df["Review"].apply(get_embedding)
````

````python
df.head()
````

### <font color ='#0F94AA'>**Complete this sentence!** <font>

````python
# Define your query
# mention your favourite type of cheese

query = "I like to eat brie" # @param {type:"string"}
````

````python
# Get the embedding for your query
query_embedding = get_embedding(query)

# Convert embeddings in DataFrame from lists to NumPy arrays if necessary
df['embedding'] = df['embedding'].apply(np.array)

# Calculate cosine similarity between the query embedding and all review embeddings
df['similarity'] = df['embedding'].apply(lambda emb: cosine_similarity([query_embedding], [emb])[0][0])

# Find the review with the highest similarity
most_similar_review = df.loc[df['similarity'].idxmax(), 'Review']
````

````python
# Find the index of the review with the highest similarity
most_similar_index = df['similarity'].idxmax()
# Get the most similar review and corresponding name
most_similar_review = df.loc[most_similar_index, 'Review']
most_similar_name = df.loc[most_similar_index, 'Name']

print("Most similar review:", most_similar_review)
print("Name associated with the review:", most_similar_name)
````

### Image parameters

- **`prompt`**

- **`n`**

- **`response_format`**
  - Description: Format of the returned response.
  - Possible Values:
    - `url`
    - `b64_json`

- **`user`**
  - Description: Unique identifier for the end-user.
  - Possible Values: Any string.

  (used for security and logging purposes at openAI)

### **Now we generate an image and retrieving it from the auto-generated URL**

````python
# Generate images with openai.Image.create()
response = client.images.generate(
    model="dall-e-3",
    prompt="an empire made of cheese,3d",
    quality="standard",
)

# Retrieve the image from a URL & display
img_bytes = get(response.data[0].url).content ## we get the image from thr url here
img = Image.open(BytesIO(img_bytes))
display(img)
````

### **Another approach is thedirect method wherein we recieve the image/response directly, not via a URL**


````python
# Return generated image directly with response_format="b64_json"
response = client.images.generate(
     model="dall-e-3",
    prompt="A panda arranging a cheese platter,watercolour",
    quality="standard",
    response_format="b64_json"
)

# Decompress image & display
img_bytes = b64decode(response.data[0].b64_json)
img = Image.open(BytesIO(img_bytes))
display(img)
````

## 💡 <font color ='#0FAA82'> **DALL-E 3 Enhanced Parameters** </font>

DALL-E 3 includes new parameters for better control over image generation. Here are the key enhancements:


### **Key DALL-E 3 Parameters:**

- **`style`**: Choose between `"vivid"` (hyper-real, dramatic) or `"natural"` (realistic, less hyper-real)
- **`quality`**: Choose between `"standard"` or `"hd"` (higher quality, costs more)
- **`revised_prompt`**: DALL-E 3 automatically enhances your prompt - you can see what it changed!

Let's see these in action:


````python
# Example 1: Using "vivid" style for dramatic, hyper-real images
response = client.images.generate(
    model="dall-e-3",
    prompt="a majestic cheese castle in the clouds",
    size="1024x1024",
    quality="standard",
    style="vivid"  # 💡 Dramatic, hyper-real style
)

# 💡 Show how DALL-E 3 enhanced the prompt
print("="*60)
print("YOUR ORIGINAL PROMPT:")
print("a majestic cheese castle in the clouds")
print("\n" + "="*60)
print("DALL-E 3 REVISED PROMPT:")
print(response.data[0].revised_prompt if hasattr(response.data[0], 'revised_prompt') else "Not available")
print("="*60 + "\n")

# Display the image
img_bytes = get(response.data[0].url).content
img = Image.open(BytesIO(img_bytes))
display(img)
````

````python
# Example 2: Using "natural" style for more realistic images
response = client.images.generate(
    model="dall-e-3",
    prompt="a cheese shop in Paris, photograph style",
    size="1024x1024",
    quality="standard",
    style="natural"  # 💡 More realistic, less dramatic
)

# Show revised prompt
print("="*60)
print("YOUR ORIGINAL PROMPT:")
print("a cheese shop in Paris, photograph style")
print("\n" + "="*60)
print("DALL-E 3 REVISED PROMPT:")
print(response.data[0].revised_prompt if hasattr(response.data[0], 'revised_prompt') else "Not available")
print("="*60 + "\n")

# Display the image
img_bytes = get(response.data[0].url).content
img = Image.open(BytesIO(img_bytes))
display(img)
````

**Why these parameters matter:**
- **`style`** gives you artistic control over the image aesthetics
- **`revised_prompt`** transparency helps you understand what DALL-E is actually generating
- **`quality="hd"`** option available when you need higher resolution (costs more)

The prompt revision feature is particularly useful for learning what makes effective image prompts!
